## Baixando em Disco Dados da API Localidades

[Link da API para municípios](https://servicodados.ibge.gov.br/api/docs/localidades#api-Municipios-municipiosGet)

```python
# baixando arquivos em disco do IBGE de municípios
print(" >Download de dados de municípios do Brasil da API do IBGE.")

# link da API
url_ibge = "https://servicodados.ibge.gov.br/api/v1/localidades/municipios"

print("  Baixando...")
# dataframe cru em JSON
df_ibge_raw = pd.read_json(url_ibge)

print("  Transformando dados...")
# transformando em um dataframe para fazer joins com o datasus
df_municipios = pd.DataFrame({
    'code_ibge_7': df_ibge_raw['id'].astype(str),
    # remoção do dígito verificador, pois DATASUS usa 6 dígitos
    'code_ibge_6': df_ibge_raw['id'].astype(str).str[:6],
    'nome_municipio': df_ibge_raw['nome'],
    'uf': df_ibge_raw.apply(
        lambda row: (
            row['microrregiao']['mesorregiao']['UF']['sigla']
            if isinstance(row['microrregiao'], dict)
            else row['regiao-imediata']['regiao-intermediaria']['UF']['sigla']
        ),
        axis=1
    )
})

# salvando em parquet
print("  Salvando em .parquet...")
pasta_municipio_ibge = pasta_projeto / 'data' / 'ibge_municipios'
pasta_municipio_ibge.mkdir(parents=True, exist_ok=True)
df_municipios.to_parquet(
    pasta_municipio_ibge / 'ibge_municipios.parquet', 
    engine='pyarrow', 
    compression='snappy'
    )

print(f"  Total de municípios carregados: {len(df_municipios)}")
```

## Arquivos de Descrição das Colunas

Vou investigar como baixar os dados que descrevem as colunas que tenho interesse. Tenho que verificar como fazer isso em FTP e como baixar através do Python para deixar disponível no disco do pc.

```python
# dicionário de dados
print(" >Download de dicionário de dados do servidor FPT...")

# pacote para baixar arquivos do servidor FPT
from ftplib import FTP
# site onde está o arquivo com dicionários completo
# ftp://ftp.datasus.gov.br/dissemin/publicos/SIHSUS/200801_/Auxiliar/TAB_SIH.zip

# acessando diretório do governo
endereco_ftp = 'ftp.datasus.gov.br'
ftp = FTP(endereco_ftp, timeout=30.0)
print(f"  Logando em {endereco_ftp}")
print(f"  {ftp.login()}")

# mensagem de boas-vindas
print("\n  Mensagem de boas-vindas")
print(f"  {ftp.getwelcome()}\n")

# identificando caminho
caminho_exato = "/dissemin/publicos/SIHSUS/200801_/Auxiliar/"
print(f"  Arquivos em {caminho_exato}\n")

# entrando no diretório exato
ftp.cwd(caminho_exato)

# loop para listagem de arquivos
ftp.retrlines('LIST')

# BAIXANDO ARQUIVIO TAB_SIH.zip
pasta_tab_sih = pasta_projeto / 'data' / 'tab_sih'
pasta_tab_sih.mkdir(parents=True, exist_ok=True)
zip_tab_sih = pasta_tab_sih / 'TAB_SIH.zip'

print("\n  Transferindo arquivo para disco local...")
with open(zip_tab_sih, 'wb') as arquivo_dicio:
    print(f"  {ftp.retrbinary('RETR TAB_SIH.zip', arquivo_dicio.write)}")

print("\n  Fechando conexão")
print(f"  {ftp.quit()}")
```

```python
# fazendo unzip do arquivo baixado

print("Fazendo unzip do arquivo baixado.")

# diretório dentro de data exclusivo para os arquivos extraídos
extracted_tab_sih = pasta_projeto / 'data' / 'extracted_tab_sih'

# verifica se o arquivo já existe e tem arquivos ou diretórios
if extracted_tab_sih.exists() and any(extracted_tab_sih.iterdir()):
    print(f"Unzip já executado em TAB_SIH.zip.\n")
else:
    print("Unzipping...")
    extracted_tab_sih.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(zip_tab_sih, 'r') as zip_ref:
        zip_ref.extractall(extracted_tab_sih)
    print("Feito.\n")
```

```python
# utilização de teste para extração de dicionário da coluna ESPEC
print("Leitura teste para coluna ESPEC (Especialidade).")
print("Arquivo consta como LEITOS.CNV")

# lista de arquivos com dados dos dicionários
dicios_list = ['LEITOS.CNV', 'IDENT.CNV', 'MARCAUTI.CNV']

# caminho para a pasta CNV, que contém dicionários
caminho_cnv = extracted_tab_sih / 'CNV' 

# arquivo cnv
arquivo_cnv = caminho_cnv / dicios_list[2]

# padrão em regex para obter dados
PATTERN_CNV = re.compile(
    r'^\s*\d+\s+(?:(?P<prefixo>\d+)-)?(?P<descricao>.+?)\s+(?P<codigo>[\d\-]+)\s*$'
)

dados = []

with open(arquivo_cnv, 'r', encoding='iso-8859-1') as f:
    for linha in f:
        linha_limpa = linha.strip()
        
        # Ignora linhas vazias e o cabeçalho inicial 'N 41 2 L'
        #if not linha_limpa or not linha_limpa[0].isdigit():
        #    continue
            
        match = PATTERN_CNV.match(linha)
        if match:
            codigo = match.group('codigo').zfill(2)
            descricao = match.group('descricao').strip()
            
            dados.append({
                'codigo_espec': codigo,
                'nome_especialidade': descricao
            })
            
df_espec = pd.DataFrame(dados)
df_espec
```
